```python
import os
import shutil

base_dir = "/content/drive/MyDrive/Combined_Dataset"
datasets = {
    "ACRIMA": "/content/drive/MyDrive/ACRIMA/ACRIMA/NOT PARTITIONED",
    "RIM-ONE": "/content/drive/MyDrive/RIM-ONE/RIM-ONE/NOT PARTITIONED",
    "Drishti-GS1": "/content/drive/MyDrive/Drishti-GS1_files"
}

glaucoma_dir = os.path.join(base_dir, "glaucoma")
normal_dir = os.path.join(base_dir, "normal")

os.makedirs(glaucoma_dir, exist_ok=True)
os.makedirs(normal_dir, exist_ok=True)
```

```python
for dataset, path in datasets.items():
    glaucoma_src = os.path.join(path, "glaucoma")
    normal_src = os.path.join(path, "normal")

    if os.path.exists(glaucoma_src):
        for img in os.listdir(glaucoma_src):
            shutil.copy(os.path.join(glaucoma_src, img), os.path.join(glaucoma_dir, f"{dataset}_{img}"))

    if os.path.exists(normal_src):
        for img in os.listdir(normal_src):
            shutil.copy(os.path.join(normal_src, img), os.path.join(normal_dir, f"{dataset}_{img}"))

print("Datasets successfully merged.")
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, Input
import os


train_dir_1 = "/content/drive/MyDrive/Drishti-GS1_files/Drishti-GS1_files/Training"
train_dir_2 = "/content/drive/MyDrive/RIM-ONE/RIM-ONE/NOT PARTITIONED"

val_dir_1 = "/content/drive/MyDrive/Drishti-GS1_files/Drishti-GS1_files/Test"
val_dir_2 = "/content/drive/MyDrive/RIM-ONE/RIM-ONE/PARTITIONED"


img_size = (128, 128)
batch_size = 32


train_datagen = ImageDataGenerator(rescale=1.0 / 255)
val_datagen = ImageDataGenerator(rescale=1.0 / 255)

def create_generator(datagen, dir1, dir2):
    gen1 = datagen.flow_from_directory(dir1, target_size=img_size, batch_size=batch_size, class_mode="binary")
    gen2 = datagen.flow_from_directory(dir2, target_size=img_size, batch_size=batch_size, class_mode="binary")
    return gen1, gen2


train_gen1, train_gen2 = create_generator(train_datagen, train_dir_1, train_dir_2)
val_gen1, val_gen2 = create_generator(val_datagen, val_dir_1, val_dir_2)

def combine_generators(gen1, gen2):
    while True:
        batch1 = next(gen1)
        batch2 = next(gen2)
        images = tf.concat([batch1[0], batch2[0]], axis=0)
        labels = tf.concat([batch1[1], batch2[1]], axis=0)
        yield images, labels

train_gen = combine_generators(train_gen1, train_gen2)
val_gen = combine_generators(val_gen1, val_gen2)


model = Sequential([
    Input(shape=(128, 128, 3)),
    Conv2D(32, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(1, activation='sigmoid')
])


model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])


model.summary()


train_samples_1 = sum([len(files) for _, _, files in os.walk(train_dir_1)])
train_samples_2 = sum([len(files) for _, _, files in os.walk(train_dir_2)])
val_samples_1 = sum([len(files) for _, _, files in os.walk(val_dir_1)])
val_samples_2 = sum([len(files) for _, _, files in os.walk(val_dir_2)])


steps_per_epoch = 8


validation_steps = (val_samples_1 + val_samples_2) // batch_size


history = model.fit(
    train_gen,
    epochs=50,
    steps_per_epoch=steps_per_epoch,
    validation_data=val_gen,
    validation_steps=validation_steps
)

```
