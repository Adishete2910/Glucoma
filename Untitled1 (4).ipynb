```python
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.image import load_img, img_to_array
import numpy as np
import os

# --- Load the model ---
model_path = "/content/glaucoma_detection_model_updated (3).h5"
if not os.path.exists(model_path):
    raise FileNotFoundError(f"Model file not found at {model_path}")

loaded_model = load_model(model_path)
print("✅ Model loaded successfully!")

# --- Function to preprocess image ---
def preprocess_image(img_path):
    if not os.path.exists(img_path):
        raise FileNotFoundError(f"Image file not found at {img_path}")
    img = load_img(img_path, target_size=(128, 128))
    img_array = img_to_array(img) / 255.0
    img_array = np.expand_dims(img_array, axis=0)
    return img_array

# --- Provide paths for both images ---
img_path1 = "/content/Im310_g_ACRIMA.jpg"
img_path2 = "/content/Im312_g_ACRIMA.jpg"

# --- Preprocess both images ---
img1 = preprocess_image(img_path1)
img2 = preprocess_image(img_path2)

# --- Predict on both images ---
pred1 = loaded_model.predict(img1)[0][0]
pred2 = loaded_model.predict(img2)[0][0]

print(f"Prediction 1 (Image 1): {pred1:.4f}")
print(f"Prediction 2 (Image 2): {pred2:.4f}")

# --- Combine predictions (average for stability) ---
final_pred = np.mean([pred1, pred2])
print(f"\nCombined Prediction Probability: {final_pred:.4f}")

# --- Final result ---
if final_pred > 0.5:
    print("🩺 Result: Glaucoma Detected")
else:
    print("🟢 Result: No Glaucoma Detected")
```
